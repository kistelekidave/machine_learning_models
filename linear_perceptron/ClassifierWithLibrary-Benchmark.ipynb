```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import numpy as np
from datasets import load_dataset


def image_preprocess(example):
    img = np.array(example["image"]) / 255.0
    img = (img > 0.5).astype(int)
    example["input"] = img.flatten()
    return example

dataset = load_dataset("ylecun/mnist")
dataset = dataset.map(image_preprocess)

full_train_data = dataset["train"]
full_test_data = dataset["test"]

X_train = np.asarray(full_train_data["input"], dtype=float)
y_train = np.asarray(full_train_data["label"], dtype=float)
X_test = np.asarray(full_test_data["input"], dtype=float)
y_test = np.asarray(full_test_data["label"], dtype=float)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
predictions = model.predict(X_test)

print(classification_report(y_test, predictions))
```
